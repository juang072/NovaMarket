## Etapa 2: Tratamiento de valores atípicos y ruido

```python
import pandas as pd

df = pd.read_csv("S06_PD_Gonzalez_DatasetDepurado.csv")
```

```python
Q1 = df["monto_compra"].quantile(0.25)
Q3 = df["monto_compra"].quantile(0.75)
IQR = Q3 - Q1
```

```python
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

print("limite inferior:", limite_inferior)
print("limite superior:", limite_superior)
```

```python
limite_inferior = 1
```

```python
outliers_iqr = df[(df["monto_compra"] < limite_inferior) | (df["monto_compra"] > limite_superior)]
```

```python
media = df["monto_compra"].mean()
desviacion = df["monto_compra"].std()
```

```python
df["z_monto_compra"] = (df["monto_compra"] - media) / desviacion
```

```python
df["es_venta_corporativa"] = df["monto_compra"] > limite_superior
```

```python
df.loc[~df["es_venta_corporativa"], "monto_compra"] = df.loc[~df["es_venta_corporativa"], "monto_compra"].clip(upper=limite_superior, lower=limite_inferior)
```

```python
ventas_diarias = df.groupby("fecha_pedido")["monto_compra"].sum().reset_index()
```

```python
ventas_diarias["suavizado"] = ventas_diarias["monto_compra"].rolling(window=7).mean()
```

```python
df.to_csv("S07_PD_Gonzalez_DatasetSinOutliers.csv", index=False)
```

## Etapa 3: Escalado de variables numéricas

## Etapa 4: Codificación de variables categóricas

```python
import pandas as pd
df = pd.read_csv("S08_PD_Gonzalez_DatasetEscalado.csv")
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
columnas_nominales = ["canal_compra", "metodo_pago", "ciudad_tienda", "categoria_producto", "comentario_cliente"]

df_codificado = pd.get_dummies(df, columns=columnas_nominales, prefix=["canal", "pago", "ciudad", "cat", "comentario"], dtype=int)
```

```python
from sklearn.preprocessing import OrdinalEncoder

orden_lealtad = ["Bronce", "Plata", "Oro", "Platino"]

codificador_lealtad = OrdinalEncoder(categories=[orden_lealtad])
df_codificado["nivel_lealtad_cod"] = codificador_lealtad.fit_transform(df_codificado[["nivel_lealtad"]])

df_codificado[["nivel_lealtad", "nivel_lealtad_cod"]].drop_duplicates()
```

```python
from sklearn.model_selection import train_test_split
```

```python
df_entrenamiento, df_prueba = train_test_split(df, test_size=0.2, random_state=42)
```

```python
from category_encoders import TargetEncoder
```

```python
codificador_ciudad = TargetEncoder(cols=["ciudad_tienda"], smoothing=10)

codificador_ciudad.fit(df_entrenamiento["ciudad_tienda"], df_entrenamiento["monto_compra"])
```

```python
df_entrenamiento["ciudad_tienda_cod"] = codificador_ciudad.transform(df_entrenamiento["ciudad_tienda"])
df_prueba["ciudad_tienda_cod"] = codificador_ciudad.transform(df_prueba["ciudad_tienda"])
```

```python
conteo_postal = df_entrenamiento["codigo_postal"].value_counts()

postales_frecuentes = conteo_postal[conteo_postal >= 15].index

df_entrenamiento["codigo_postal_agrupado"] = df_entrenamiento["codigo_postal"].where(
    df_entrenamiento["codigo_postal"].isin(postales_frecuentes), "Otras"
)

df_prueba["codigo_postal_agrupado"] = df_prueba["codigo_postal"].where(
    df_prueba["codigo_postal"].isin(postales_frecuentes), "Otras"
)
```

```python
df_entrenamiento.to_csv("S09_PD_Gonzalez_DatasetCodificado.csv", index=False)
```

## Diccionario de datos

| Columna del dataset | De qué semana viene | Qué significa |
| :--- | :--- | :--- |
| **z_monto_compra** | Semana 7 | *Z-score* del monto de compra para medir cuántas desviaciones estándar se aleja de la media. |
| **es_venta_corporativa** | Semana 7 | Indicador booleano que identifica si la compra superó el límite superior de outliers por IQR. |
| **edad_cliente_escalada** | Semana 8 | Edad del cliente estandarizada con `StandardScaler` . |
| **unidades_vendidas_escalada** | Semana 8 | Unidades vendidas escaladas mediante `MinMaxScaler`. |
| **monto_compra_escalado** | Semana 8 | Monto de compra transformado con `RobustScaler`para mitigar el efecto de atípicos. |
| **canal_\***, **pago_\***, **ciudad_\***, **cat_\*** | Semana 9 | Variables binarias (0/1) creadas por One-Hot Encoding (`get_dummies`) para representar las variables categóricas nominales. |
| **nivel_lealtad_cod** | Semana 9 | Codificación numérica ordinal del nivel de lealtad (Bronce=0, Plata=1, Oro=2, Platino=3). |
| **ciudad_tienda_cod** | Semana 9 | Codificación por objetivo (*Target Encoding*) de la ciudad con base en la media del `monto_compra`. |
| **codigo_postal_agrupado** | Semana 9 | Código postal que conserva las zonas de alta frecuencia y agrupa las demás bajo "Otras". |
| **mes_pedido** | Semana 10 | Mes del año (1 a 12) extraído de la fecha del pedido. |
| **trimestre_pedido** | Semana 10 | Trimestre del año (1 a 4) extraído de la fecha del pedido. |
| **dia_semana_pedido** | Semana 10 | Día de la semana (0 = Lunes, 6 = Domingo) extraído de la fecha. |
| **es_fin_semana_pedido** | Semana 10 | Variable binaria (1 si el pedido fue en sábado o domingo, 0 si fue día hábil). |
| **longitud_comentario** | Semana 10 | Cantidad total de caracteres presentes en el comentario del cliente. |
| **num_palabras_comentario** | Semana 10 | Número de palabras contadas en el comentario del cliente. |
| **tiene_comentario** | Semana 10 | Indicador binario (1 si el cliente escribió comentario, 0 si estaba vacío). |
| **menciona_demora** | Semana 10 | Indicador binario (1 si el comentario contiene la palabra/raíz "demor", 0 si no). |
| **grupo_edad** | Semana 10 | Categorización por rangos de edad ("18 a 25", "26 a 35", etc.) obtenida con `pd.cut`. |
| **monto_compra_log** | Semana 10 | Transformación logarítmica (`np.log1p`) aplicada al monto de compra para reducir la asimetría. |
| **monto_compra_boxcox** | Semana 10 | Transformación de potencia Box-Cox aplicada al monto de compra para normalizar la distribución. |
